## 变量与类型

Python 变量不需要提前声明。变量只是绑定到对象的名称，真正拥有类型的是对象本身。

### 变量

变量必须先赋值再使用。同一个变量名可以在不同时间绑定到不同类型的对象。

```python
counter = 100
weight = 76.5
name = "Blade"
print(counter, weight, name)

age = 100
print(age)
age = "some string"
print(age)
```

### 命名

变量名只能包含字母、数字和下划线，不能以数字开头，也不能使用 Python 关键字。命名应简洁且有描述性。

```python
a = b = c = 100
print(a, b, c)

a, b, c = 1, 2.0, "China"
print(a, b, c)
```

### 基本类型

常见标准类型包括数字、字符串、列表、元组、集合和字典。数字类型包括 `int`、`float`、`bool` 和 `complex`。

```python
a, b, c, d = 20, 5.5, True, 4 + 3j
print(type(a), type(b), type(c), type(d))
print(True + 1)
print(isinstance(a, int))
print(isinstance(True, float))
```

### 引用

变量保存的是对象引用。`id()` 可以查看对象标识。不可变对象发生“修改”时，通常会绑定到新对象；可变对象可以在原对象上改变内容。

```python
number = 3
print(hex(id(number)))
number = 5
print(hex(id(number)))

profile = {"name": "c++"}
print(hex(id(profile)))
profile["name"] = "python"
print(hex(id(profile)))
```

## 运算符

运算符用于完成算术、比较、赋值、位运算、逻辑判断、成员判断和身份判断。

### 算术运算符

常见算术运算符包括 `+`、`-`、`*`、`/`、`//`、`%` 和 `**`。`/` 返回浮点数，`//` 返回向下取整后的结果。

```python
a = 24
b = 5
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a**b)
print(-5 // 2)
print(-123 % 10)
```

### 比较和赋值

比较运算符返回布尔值。复合赋值如 `+=`、`*=` 会基于当前值计算后再赋回变量。

```python
a = 21
b = 10
print(a == b)
print(a != b)
print(a > b)

c = 0
c += a
c *= 2
c //= 3
print(c)
```

### 位运算符

位运算符会把整数按二进制位处理，包括 `&`、`|`、`^`、`~`、`<<` 和 `>>`。

```python
a = 60
b = 13
print(a & b)
print(a | b)
print(a ^ b)
print(~a)
print(a << 2)
print(a >> 2)
```

### 逻辑、成员和身份

`and`、`or`、`not` 用于逻辑判断；`in` 和 `not in` 用于成员判断；`is` 判断两个变量是否引用同一个对象，值相等应使用 `==`。

```python
items = [1, 2, 3, 4, 5]
print(2 in items)
print(10 not in items)
print(True and False)
print(True or False)

left = [1, 2, 3]
right = [1, 2, 3]
alias = left
print(left == right)
print(left is right)
print(left is alias)
```

## 字符串

字符串是字符的有序集合，可以用单引号、双引号或三引号创建。字符串不可变，支持索引、切片、拼接、重复和常用方法。

### 创建和访问

索引从 `0` 开始，负索引从末尾向前数。切片形式为 `sequence[start:end:step]`，遵循左闭右开规则。

```python
text = "hello crash course"
print(text[6:11])
print(text[:-7])
print(text[::2])
print(text[::-1])
print(len(text))
```

### 转义和多行字符串

反斜杠用于转义。原始字符串前缀 `r` 可以让反斜杠保持字面含义。三引号可创建多行字符串。

```python
print("Ro\na\tldo")
print(r"Ro\na\tldo")

sample = """
This is sample,
it is very easy.
"""
print(sample)
```

### 格式化和方法

现代代码优先使用 f-string。常见方法包括 `title()`、`upper()`、`lower()`、`split()`、`join()`、`replace()`、`startswith()` 和 `strip()`。

### 类型转换

字符串和基本类型之间通常用构造函数转换，例如 `str()`、`int()`、`float()` 和 `bool()`。注意非空字符串转布尔值总是 `True`。

```python
print(str(100))
print(str(100.1))
print(str(True))
print(bool("False"))
print(bool(""))
print(int("123"))
print(float("123.4"))
```

## 进制转换

程序中常见进制包括二进制、八进制、十进制和十六进制。Python 可以直接书写不同进制的整数字面量，也可以用内置函数转换。

### 字面量

`0b` 表示二进制，`0o` 表示八进制，`0x` 表示十六进制。

```python
print(0b1101010)
print(0o125)
print(0x3F2A1)
```

### 函数转换

`int(value, base)` 可以按指定进制解析字符串。`bin()`、`oct()` 和 `hex()` 可以把整数转成对应进制字符串。

```python
print(int("1101010", 2))
print(int("125", 8))
print(int("3f2a1", 16))

print(bin(106))
print(oct(85))
print(hex(258721))
print(bin(int("125", 8)))
print(hex(int("1101010", 2)))
```

## 列表

**学习目标：** 本节介绍列表的创建、索引切片、常用方法、遍历、嵌套列表和列表推导，重点理解可变序列的操作方式。

### 列表基础

前面讨论字符串时，我们介绍了 `Python` 中序列 `sequence` 的概念。列表是最常用的序列类型之一。与字符串不同，列表是可变对象，列表中的元素可以被修改。

本节主要讨论：

1. 创建列表。
2. 索引和切片。
3. 常用列表方法。
4. 遍历列表。
5. 嵌套列表。
6. 列表推导式。

列表由方括号\[\]和分隔列表中的每个元素的逗号组成。

```python
my_list = [1, 2, 3]
```

上面我们创建了一个全部由整数组成的列表。事实上，列表中的元素可以不同类型，列如：

```python
my_list = ["A string", 1, 3.14, 1 + 1j]
print(my_list)
```

和字符串一样，`len()` 函数可以获取列表序列中有多少项。

```python
print(len(my_list))
```

### 索引与切片

列表的索引与切片基本与字符串的操作一致。

```python
my_list = ["one", "two", "three", 4, 5]
```

```python
# 抽取第一个元素
print(my_list[0])
```

```python
# 抽取索引1开始的所有元素
print(my_list[1:])
```

```python
# 抽取从头开始到第3个元素
print(my_list[:3])
```

了解列表的切片操作之后，那么复制一个列表就非常简单了。

与字符串一样，可以通过运算符‘+’来拼接不同的列表，例如

```python
print(my_list + ["add new item"])
```

当然，上面的操作并不会改变原来列表的内容，打印 my_list 的内容，应该会看到和原来是一样的。

```python
print(my_list)
```

可以通过重新赋值的方式改变列表的值，例如：

```python
my_list = my_list + ["append a new item"]  # 等价于 += 追加元素
print(my_list)
```

类似于字符串，可以使用\*符号复制列表，例如：

```python
print(my_list * 2)
```

### 列表方法

`Python` 的列表与 `C` / `C++` 数组有相似之处，但操作更灵活。列表没有固定大小，也不要求所有元素类型一致，因此更接近动态数组或泛型容器。

```python
my_list = [1, 2, 3]
```

- **append**方法在列表末尾添加元素。

```python
my_list.append("append a new item")
print(my_list)
```

- **extend**方法通过迭代的方式在列表末尾添加元素。注意与 append 方法的区别。例如

```python
my_list = [1, 2, 3]
my_list.append([4, 5])
print(my_list)
my_list = [1, 2, 3]
my_list.extend([4, 5])
print(my_list)
```

- **count**接受一个元素并返回它在列表中出现的次数。

```python
my_list = [1, 2, 3, 4, 3, 4, 5]
print(my_list.count(3))
```

- **index**将返回作为参数放置的任何元素的索引。注意:如果元素不在列表中，则返回错误。

```python
my_list = [0, 1, 2, 3, 4]
my_list.index(2)
```

- **pop**方法删除列表元素。

使用 pop 从列表中删除一个元素。默认情况下 pop 会删除最后一个索引，但是你也可以指定删除哪个索引。例如:

```python
my_list = [1, 2, 3, 4]
my_list.pop()  # 默认删除最后一个元素
print(my_list)
```

```python
my_list.pop(0)  # 删除指定索引的元素
print(my_list)
```

- **remove** 方法会从左到右查找并删除第一个符合要求的元素。

如果列表中有多个相同元素，`remove(value)` 只会删除最靠左的第一个匹配项；如果元素不存在，会抛出 `ValueError`。

```python
my_list = [1, 2, 3, 4, 3]
my_list.remove(3)
print(my_list)  # [1, 2, 4, 3]
```

- **`sort`**和**reverse**方法分别将列表中元素排序和逆序。

```python
my_list = ["y", "e", "m", "e", "n", "g", "j", "i", "e"]
print(my_list)
```

```python
my_list.sort()
print(my_list)
```

### 遍历列表

在很多的应用场景，我们需要遍历整个列表，对每一个元素执行相同的操作。遍历列表通常常用 for 循环来实现。例如：

```python
courses = ["C++", "Python", "Java"]
for course in courses:
    print(course)
```

遍历列表需要注意的问题是:
- for 循环后面不要忘记加上冒号:
- for 循环里面的语句缩进要小心，相同的缩进表示都在循环里面。

掌握了遍历列表的方法后，我们创建列表经常使用的方法是采用 range 方法。`range()` 可以很方便地生成一系列的数字。例如：

```python
for value in range(1, 11):
    print(value, end=" ")
```

使用 `range()` 方法时，注意 `range(a,b)` 表示的区间时\[a,b)，即左闭右开区间。

另外，还要注意，`range()` 可以产生一系列数字，但它返回的类型并不是列表，可以打印一下试一试。

```python
range_list = range(1, 11)
print(type(range_list))
```

但时我们可以使用 list 方法将 range 类型转换为列表。

```python
numbers = list(range(1, 11))
print(numbers)
print(type(numbers))
```

### 嵌套列表

```python
my_list.reverse()
print(my_list)
```

`Python` 数据结构支持嵌套。这意味着可以在一种数据结构中存放另一种数据结构，例如列表中的列表。

```python
# 三个列表
lst_1 = [1, 2, 3]
lst_2 = [4, 5, 6]
lst_3 = [7, 8, 9]

# 嵌套为矩阵
matrix = [lst_1, lst_2, lst_3]
print(matrix)
```

现在我们可以再次使用索引来获取元素，但是现在索引有两个级别。矩阵对象中的项，然后是列表中的项!

```python
print(matrix[0])
print(matrix[0][0])
```

### 列表推导

列表推导式是 `Python` 中快速构建列表的常用语法。它可以把一个简单循环和可选条件写在方括号内，让列表生成逻辑更紧凑。

下面通过几个例子理解列表推导式。

#### 示例 1

```python
# 创建一个字符串
my_string = "something"
print(my_string)
# 用字符串的每一个字符构成列表
my_list = [c for c in my_string]
print(my_list)
```

#### 示例 2

```python
# 区间[0, 11)的平方数
my_list = [x**2 for x in range(0, 11)]
print(my_list)
```

#### 示例 3

```python
# 区间[0, 11)的所有偶数
my_list = [x for x in range(11) if x % 2 == 0]
print(my_list)
```

#### 示例 4

```python
# 摄氏转华氏
celsius = [0, 10, 20.1, 34.5]

fahrenheit = [((float(9) / 5) * temp + 32) for temp in celsius]

print(fahrenheit)
```

#### 示例 5

使用嵌套推导

```python
my_list = [x**2 for x in [x**2 for x in range(0, 11)]]
print(my_list)
```

## 元组

**学习目标：** 本节介绍元组的创建和常用操作，重点比较元组与列表在可变性和使用场景上的差异。

### 元组基础

在 `Python` 中，元组 `tuple` 与列表相似，但元组是不可变对象，创建后不能修改其中的元素。

元组适合表示不应被修改的数据，例如图像尺寸、日期片段、坐标点或函数返回的多项结果。

本节主要讨论：

1. 元组创建方法。
2. 元组常用操作。
3. 元组与列表的差异。

### 元组创建

元组使用圆括号 `()` 表示，元素之间用逗号分隔。

```python
tuple_1 = (1, 2, 3)
tuple_2 = ("python", 100)
tuple_3 = "a", "b", "c", 1, 2, 3
print(tuple_1)
print(tuple_2)
print(tuple_3)
```

当然，也可以创建空元组，如：

```python
empty_tuple = tuple()
print(empty_tuple)
```

但是，当创建的元组只有一个元素时，需要加上逗号，例如：

```python
single_tuple = (1,)
print(single_tuple)
print(type((1)))
print(type((1,)))
```

`len()` 函数可以获取元组有多少项。

```python
my_tuple = (1, 2, 3)
print(len(my_tuple))
```

### 元组操作

元组操作也有很多内置函数，当然，没有列表那么多操作方法。下面介绍几个常见的方法。

- 访问元组
元组可以使用下标索引来访问元组中的值，如：

```python
tuple_1 = ("Python", "C++", 60, 59)
tuple_2 = (1, 2, 3, 4, 5, 6, 7)
print(tuple_1[0])
print(tuple_2[3:6])
```

- 修改元组
元组中的元素值是不允许修改的，但我们可以对元组进行连接组合，如:

```python
tuple_1 = (12, 34.56)
tuple_2 = ("abc", "xyz")

# 以下修改元组元素操作是非法的。
# tuple_1[0] = 100

# 创建一个新的元组
tuple_3 = tuple_1 + tuple_2
print(tuple_3)
```

- 删除元组
元组中的元素值是不允许删除的，但我们可以使用 del 语句来删除整个元组，如:

```python
tuple_1 = ("Python", "C++", 60, 59)
print(tuple_1)
del tuple_1
try:
    print(tuple_1)
except NameError as exc:
    print(type(exc).__name__)
```

- 元素索引
通过 index 方法可以获取元素在元组中的索引，如：

```python
tuple_1 = ("Python", "C++", 60, 59)
ind = tuple_1.index("C++")
print(ind)
```

- 元素个数统计
通过 count 方法可以统计元组中元素出现的次数，如：

```python
tuple_1 = (1, 2, 3, 4, 4, 4, 3, 3, 2, 1)
cnt = tuple_1.count(4)
print(cnt)
```

### 小结

在实际开发中，元组不像列表那样经常使用，但是在需要不可变性时才会使用。如果在程序中传递一个对象并需要确保它不会被更改，使用 tuple 会比较方便。

## 字典

字典是键值对集合，用于存储映射关系。键必须是可哈希对象，值可以是任意对象。字典是可变对象。

### 创建和访问

字典可以用花括号、`dict()` 或字典推导式创建。

```python
my_dict = {"key1": "value1", "key2": "value2"}
print(my_dict)
print(my_dict["key2"])

pairs = dict([(1, "a"), (2, "b"), (3, "c")])
print(pairs)

squares = {x: x * x for x in range(20) if x % 2 == 1}
print(squares)
```

### 修改和删除

字典可以修改已有键的值，也可以通过赋值新增键。删除键可用 `del` 或 `pop()`，清空字典可用 `clear()`。

```python
profile = {"name": "Zara", "age": 7}
profile["age"] = 8
profile["class"] = "First"
print(profile)
print(profile.pop("class"))
print(profile)
```

### 遍历

`items()` 遍历键值对，`keys()` 遍历键，`values()` 遍历值。需要顺序时可以配合 `sorted()`。

```python
favorite_languages = {"jen": "python", "sarah": "c", "edward": "ruby", "phil": "python"}
for name, language in favorite_languages.items():
    print(f"{name}: {language}")

for name in sorted(favorite_languages.keys()):
    print(name.title())

for language in set(favorite_languages.values()):
    print(language.title())
```

### 常用方法

常用方法包括 `copy()`、`get()`、`items()`、`keys()`、`pop()`、`popitem()`、`update()` 和 `values()`。

```python
dict1 = {1: "Python", 2: "Java", 3: "Ruby", 4: "Scala"}
dict2 = dict1.copy()
print(dict2.get(1))
print(dict2.items())
print(dict2.keys())
dict2.pop(4)
dict2.update({3: "Scala"})
print(dict2.values())
```

## 集合

集合 `set` 是无序、可迭代、可变且不包含重复元素的数据类型。它常用于去重、成员判断和数学集合运算。

### 创建

集合可以用花括号或 `set()` 创建。空花括号 `{}` 创建的是字典，空集合应使用 `set()`。

```python
print({1, 2, 3})
print({1.0, "Hello", (1, 2, 3)})
print(set("python"))
print(set([1, 2, 4, 4, 3, 3, 3]))
print(type({}))
print(type(set()))
```

### 修改和删除

`add()` 添加单个元素，`update()` 添加多个元素。`discard()` 删除不存在元素时不报错，`remove()` 删除不存在元素时会抛出 `KeyError`。

```python
my_set = set()
my_set.add(8)
my_set.update([10, 11])
print(my_set)

my_set.discard(8)
print(my_set)
my_set.clear()
print(my_set)
```

### 集合运算

集合支持并集、交集、差集和对称差集，对应运算符是 `|`、`&`、`-` 和 `^`。

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
print(A | B)
print(A & B)
print(A - B)
print(A ^ B)
```

### frozenset

`frozenset` 是不可变集合，可哈希，因此可以作为字典键或另一个集合的元素。

```python
A = frozenset([1, 2, 3, 4])
B = frozenset([3, 4, 5, 6])
print(A.isdisjoint(B))
print(A.intersection(B))
print(frozenset({1, 2, 3, 4, 5, 6}))
```

## 函数定义

函数把一组语句封装成可复用逻辑，可以接收参数，也可以返回结果。函数体以 `def` 开头，并通过缩进确定范围。

### 基本结构

函数第一行可以写文档字符串，用于说明用途、参数和返回值。

### 传递列表

列表是可变对象。函数内部修改传入列表会影响调用方。如果要保留原列表，可以传入切片副本。

```python
def print_models(unprinted, completed):
    while unprinted:
        current = unprinted.pop()
        completed.append(current)


unprinted_designs = ["iphone case", "robot pendant", "dodecahedron"]
completed_models = []
print_models(unprinted_designs[:], completed_models)
print(unprinted_designs)
print(completed_models)
```

### global 和 nonlocal

内部作用域要修改全局变量时使用 `global`。内部函数要修改外层非全局作用域变量时使用 `nonlocal`。

```python
num = 1


def update_global():
    global num
    num = 123


update_global()
print(num)


def outer():
    value = 10

    def inner():
        nonlocal value
        value = 100

    inner()
    return value


print(outer())
```

## Lambda 与 map

`map()` 把函数惰性地应用到一个或多个可迭代对象上。需要列表时显式调用 `list()`；简单转换通常优先使用列表推导。

### divmod

`divmod(a, b)` 同时返回商和余数，结果等价于 `(a // b, a % b)`。它适合分页、时间换算、进制拆分这类需要同时拿到两部分结果的场景。

```python
total_seconds = 3_725
minutes, seconds = divmod(total_seconds, 60)
hours, minutes = divmod(minutes, 60)
print(f"{hours:02d}:{minutes:02d}:{seconds:02d}")

page_size = 20
total_items = 95
full_pages, remainder = divmod(total_items, page_size)
page_count = full_pages + bool(remainder)
print(page_count)
```

### map、filter 和 reduce

`map()` 和 `filter()` 返回惰性迭代器。`reduce()` 会反复应用二元函数并返回最终值；若能用 `sum()`、`max()`、`min()` 或清晰循环表达，通常优先选择这些写法。

## 生成器与迭代器

容器把元素组织在一起，迭代器按需返回下一个值，生成器是一种特殊迭代器。理解迭代协议有助于处理大数据序列和流式数据。

### 生成器表达式

列表推导式会一次性创建列表，生成器表达式会按需产生元素。

```python
lst = [x * x for x in range(5)]
generator = (x * x for x in range(5))
print(lst)
print(generator)
print(next(generator))
print(next(generator))
```

### 生成器函数

包含 `yield` 的函数会返回生成器对象。执行到 `yield` 时函数暂停，下一次迭代从暂停位置继续。

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b


print(list(fibonacci(10)))
```

```python
from collections.abc import Iterable, Iterator


values = [1, 2, 3]
iterator = iter(values)
print(isinstance(values, Iterable))
print(isinstance(values, Iterator))
print(isinstance(iterator, Iterator))
print(next(iterator))
```

## math

`math` 面向实数计算，提供三角函数、组合数、最大公约数、精确求和、距离计算等能力。它适合普通工程计算，但涉及金额和精确小数时不要用 `float` 直接计算。

```python
import math


def distance(point_a: tuple[float, ...], point_b: tuple[float, ...]) -> float:
    return math.dist(point_a, point_b)


def stable_average(values: list[float]) -> float:
    if not values:
        raise ValueError("values must not be empty")
    return math.fsum(values) / len(values)


print("hypot:", math.hypot(3, 4))
print("distance:", distance((0, 0), (3, 4)))
print("average:", stable_average([0.1, 0.2, 0.3]))
print("is close:", math.isclose(0.1 + 0.2, 0.3, rel_tol=1e-9))
print("gcd/lcm:", math.gcd(12, 18), math.lcm(12, 18))
print("comb/perm:", math.comb(5, 2), math.perm(5, 2))
```

### 常用选择

| 需求 | 推荐 API | 说明 |
| --- | --- | --- |
| 判断浮点数近似相等 | `math.isclose()` | 不要直接用 `==` 比较浮点计算结果 |
| 多个浮点数求和 | `math.fsum()` | 比内置 `sum()` 更适合浮点误差敏感场景 |
| 坐标距离 | `math.dist()` / `math.hypot()` | 比手写平方和更清晰 |
| 组合排列 | `math.comb()` / `math.perm()` | 不需要自己写阶乘公式 |
| 整数约分辅助 | `math.gcd()` / `math.lcm()` | 常用于分数、周期和步长计算 |

## cmath

`cmath` 面向复数计算。即使输入是普通数字，它的返回值也通常是复数。信号处理、阻抗计算、复平面旋转和复数方程根可以优先考虑它。

```python
import cmath
import math as real_math


def impedance(resistance: float, reactance: float) -> complex:
    return complex(resistance, reactance)


def rotate(point: complex, radians: float) -> complex:
    return point * cmath.exp(1j * radians)


z = impedance(3, 4)
radius, angle = cmath.polar(z)
restored = cmath.rect(radius, angle)
roots = [cmath.exp(2j * real_math.pi * index / 3) for index in range(3)]

print("z:", z)
print("abs:", abs(z))
print("phase:", cmath.phase(z))
print("polar/restored:", (radius, angle), restored)
print("rotated:", rotate(1 + 0j, real_math.pi / 2))
print("third roots of 1:", roots)
```

## decimal

`decimal.Decimal` 适合金额、结算、报表和需要明确舍入规则的小数计算。关键原则是：从字符串创建 `Decimal`，不要从浮点字面量创建。

```python
from decimal import Decimal, ROUND_HALF_UP, localcontext

CENT = Decimal("0.01")


def money(value: str) -> Decimal:
    return Decimal(value).quantize(CENT, rounding=ROUND_HALF_UP)


def apply_tax(amount: Decimal, tax_rate: Decimal) -> Decimal:
    return (amount * (Decimal("1") + tax_rate)).quantize(CENT, rounding=ROUND_HALF_UP)


subtotal = sum((money("19.995"), money("5.235"), money("3.10")), Decimal("0.00"))
total = apply_tax(subtotal, Decimal("0.06"))

with localcontext() as ctx:
    ctx.prec = 8
    ratio = Decimal("1") / Decimal("7")

print("float issue:", 0.1 + 0.2)
print("decimal sum:", Decimal("0.1") + Decimal("0.2"))
print("subtotal:", subtotal)
print("total:", total)
print("local precision:", ratio)
```

## fractions.Fraction

`Fraction` 表示有理数，适合比例、精确分数、配方换算和需要避免浮点误差的计算。它会自动约分，也可以用 `limit_denominator()` 把小数近似成易读分数。

```python
from fractions import Fraction


def completion_ratio(done: int, total: int) -> Fraction:
    if total <= 0:
        raise ValueError("total must be positive")
    return Fraction(done, total)


one_third = Fraction(1, 3)
recipe = {"flour": Fraction(3, 2), "sugar": Fraction(1, 4)}
scaled_recipe = {name: amount * 2 for name, amount in recipe.items()}
approx_pi = Fraction(3.1415926).limit_denominator(1000)

print("1/3 + 1/6:", one_third + Fraction(1, 6))
print("completion:", completion_ratio(45, 120))
print("scaled recipe:", scaled_recipe)
print("decimal string:", Fraction("0.125"))
print("pi approx:", approx_pi)
```

## calendar

`calendar` 用于生成和分析日历数据，适合月历展示、闰年判断、按周遍历日期、统计工作日和生成简单 HTML 日历。它只处理日历结构，不负责时区、时间点和夏令时；这些场景应结合 `datetime` 或 `zoneinfo`。

```python
import calendar
from datetime import date


def month_overview(year: int, month: int) -> dict[str, int | bool]:
    first_weekday, days_in_month = calendar.monthrange(year, month)
    return {
        "first_weekday": first_weekday,
        "days_in_month": days_in_month,
        "is_leap_year": calendar.isleap(year),
    }


print(calendar.month(2026, 5))
print("Overview:", month_overview(2026, 5))
print("Leap days 2020-2030:", calendar.leapdays(2020, 2031))
print("Today weekday:", calendar.day_name[date.today().weekday()])
```

### 按周遍历月份

`Calendar.monthdatescalendar()` 会返回完整周矩阵，矩阵中可能包含上个月或下个月的日期。这个结构很适合渲染日历组件或做按周统计。

```python
import calendar as calendar_tools
from datetime import date


def month_weeks(
    year: int, month: int, first_weekday: int = calendar_tools.MONDAY
) -> list[list[date]]:
    cal = calendar_tools.Calendar(firstweekday=first_weekday)
    return cal.monthdatescalendar(year, month)


for week in month_weeks(2026, 5):
    print([day.isoformat() for day in week])
```

### 工作日统计

`itermonthdates()` 适合按日期流式遍历一个月。统计工作日时要过滤掉矩阵中属于相邻月份的日期，再排除周末和节假日。

```python
import calendar as business_calendar
from datetime import date


def business_days(
    year: int, month: int, holidays: set[date] | None = None
) -> list[date]:
    holidays = holidays or set()
    cal = business_calendar.Calendar(firstweekday=business_calendar.MONDAY)
    days: list[date] = []

    for day in cal.itermonthdates(year, month):
        if day.month != month:
            continue
        if day.weekday() >= 5:
            continue
        if day in holidays:
            continue
        days.append(day)

    return days


may_holidays = {date(2026, 5, 1)}
workdays = business_days(2026, 5, may_holidays)
print("Business days:", len(workdays))
print("First five:", workdays[:5])
```

## 目录操作

目录操作包括创建目录、列出目录内容、筛选文件和子目录，以及递归遍历。现代 Python 推荐优先使用 `pathlib.Path`。

### 创建示例目录

下面创建一个小型目录树，便于演示遍历操作。

```python
from pathlib import Path


base = Path("my_directory")
for directory in [base / "sub", base / "logs", base / "config"]:
    directory.mkdir(parents=True, exist_ok=True)

for file_path in [
    base / "file1.py",
    base / "file2.csv",
    base / "file3.txt",
    base / "sub" / "bar.py",
    base / "sub" / "foo.py",
]:
    file_path.parent.mkdir(parents=True, exist_ok=True)
    file_path.touch(exist_ok=True)

print(sorted(entry.name for entry in base.iterdir()))
```

### iterdir 和 scandir

`Path.iterdir()` 返回目录下的 `Path` 对象。`os.scandir()` 返回带文件类型信息的目录项，适合较底层扫描。

```python
import os


for entry in base.iterdir():
    print(entry.name)

with os.scandir(base) as entries:
    for entry in entries:
        print(entry.name, entry.is_file(), entry.is_dir())
```

## 构造与资源管理

`__init__()` 在实例创建后初始化对象状态。`__new__()` 负责创建实例，大多数普通类只需要实现 `__init__()`。

### __init__

创建实例时，Python 会自动调用 `__init__()`。

```python
class Person:
    def __init__(self, name):
        self.name = name

    def print_name(self):
        print(self.name)


person = Person("yemengjie")
person.print_name()
```

### __new__

调用顺序是先执行 `__new__()`，再执行 `__init__()`。只有在控制实例创建过程或继承不可变类型等特殊场景中，才需要自定义 `__new__()`。

```python
class Person:
    def __new__(cls, name: str):
        print("call __new__")
        return super().__new__(cls)

    def __init__(self, name: str) -> None:
        print("call __init__")
        self.name = name


person = Person("yemengjie")
print(person.name)
```

### 上下文管理器

`__del__()` 调用时机不确定，不适合管理关键资源。文件、网络连接、锁等资源更推荐用 `with` 和上下文管理协议管理。

```python
class ManagedResource:
    def __enter__(self):
        print("open resource")
        return self

    def __exit__(self, exc_type, exc, traceback):
        print("close resource")
        return False

    def use(self):
        print("use resource")


with ManagedResource() as resource:
    resource.use()
```

## 属性与方法

属性保存状态，方法描述行为。属性可分为实例属性和类属性，方法可分为实例方法、类方法和静态方法。

### 实例属性和类属性

实例属性属于某个对象，类属性由所有实例共享。

```python
class Person:
    species = "human"

    def __init__(self, name="no one"):
        self.name = name


p1 = Person()
p2 = Person("yemengjie")
print(p1.species, p1.name)
print(p2.species, p2.name)
```

### 实例方法、类方法和静态方法

实例方法需要实例状态，类方法通过 `cls` 访问类状态，静态方法不依赖类或实例状态。

```python
class Person:
    species = "human"

    def __init__(self, name="no one"):
        self.name = name

    def instance_method(self):
        print("instance method", self.name)

    @classmethod
    def class_method(cls):
        print("class method", cls.species)

    @staticmethod
    def static_method():
        print("static method")


person = Person()
person.instance_method()
Person.class_method()
Person.static_method()
```

### 公开、受保护和私有

一个下划线开头表示受保护成员约定，两个下划线开头表示私有成员。私有成员不应在类外直接访问。

```python
class Person:
    def __init__(self, name="no one", age=18, address="bnuz"):
        self.name = name
        self.__age = age
        self._address = address

    def show_profile(self):
        print(f"name: {self.name}, age: {self.__age}, address: {self._address}")


person = Person()
person.show_profile()
print(person.name)
print(person._address)
```

## 特殊方法

特殊方法让自定义对象支持字符串表示、索引、长度、调用、比较和运算符等 Python 协议。

### 字符串表示

`__repr__()` 返回面向开发者的表示，`__str__()` 返回面向用户的表示。如果没有 `__str__()`，通常会使用 `__repr__()`。

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __str__(self):
        return f"({self.x}, {self.y})"


v = Vector(3, 4)
print(repr(v))
print(str(v))
```

### 运算符重载

实现 `__add__()`、`__sub__()`、`__eq__()`、`__lt__()` 等方法，可以让对象支持 `+`、`-`、`==`、`<` 等运算符。

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __len__(self):
        return int(self.x**2 + self.y**2)

    def __eq__(self, other):
        return len(self) == len(other)

    def __lt__(self, other):
        return len(self) < len(other)


v1 = Vector(1, 1)
v2 = Vector(2, 2)
print(f"{v1} + {v2} = {v1 + v2}")
print(f"{v1} - {v2} = {v1 - v2}")
print(v1 == v2)
print(v1 < v2)
```

## 继承

继承让子类复用父类属性和方法，也可以在父类基础上扩展新行为。继承适合表达 `is-a` 关系；如果只是复用能力，组合往往更清晰。

### 单继承

子类可以显式调用父类构造函数，也可以用 `super()` 调用方法解析顺序中的下一个实现。

```python
class Person:
    def __init__(self, name, age, weight):
        self.name = name
        self.age = age
        self.__weight = weight

    def show_basic_info(self):
        print(f"{self.name}: age={self.age}, weight={self.__weight} kg")


class Student(Person):
    def __init__(self, name, age, weight, grade):
        super().__init__(name, age, weight)
        self.grade = grade

    def show_student_info(self):
        print(f"{self.name}: age={self.age}, grade={self.grade}")


student = Student("ken", 10, 30, 3)
student.show_basic_info()
student.show_student_info()
```

### 多继承和组合

多继承会增加方法解析顺序和初始化复杂度。实际项目中应谨慎使用，优先选择组合或更清晰的继承结构。

```python
class Speaker:
    def __init__(self, name, topic):
        self.name = name
        self.topic = topic

    def speak(self):
        print(f"I'm {self.name}, my topic is {self.topic}")


class Sample:
    def __init__(self, name, topic):
        self.speaker = Speaker(name, topic)


test = Sample("Tim", "Python")
test.speaker.speak()
print(Sample.mro())
```

### 方法重写

子类可以定义同名方法覆盖父类实现。`super()` 可用于调用父类或 MRO 中的下一个类的方法。

```python
class Parent:
    def my_method(self):
        print("调用父类方法")


class Child(Parent):
    def my_method(self):
        print("调用子类方法")
        super().my_method()


child = Child()
child.my_method()
```

### 子类构造函数

子类定义自己的 `__init__()` 后，父类 `__init__()` 不会自动执行。如果需要父类初始化逻辑，应在子类构造函数中调用 `super().__init__(...)`。

```python
class Father:
    def __init__(self, name):
        self.name = name
        print(f"father name: {self.name}")

    def get_name(self):
        return f"Father {self.name}"


class Son(Father):
    def __init__(self, name):
        super().__init__(name)
        print("Son init")

    def get_name(self):
        return f"Son {self.name}"


son = Son("morningstar")
print(son.get_name())
```

## collections

`collections` 是标准库中的容器工具模块。这个主题重点介绍 `Counter` 和 `defaultdict`。

### Counter

`Counter` 用于统计元素出现次数，结果像一个字典，键是元素，值是计数。

```python
from collections import Counter


letters = Counter("blue")
words = Counter(["red", "blue", "red", "green", "blue", "blue"])
print(letters)
print(words)
print(words.most_common(2))
```

### defaultdict

`defaultdict` 是 `dict` 的子类。访问缺失键时，它会调用工厂函数创建默认值，而不是抛出 `KeyError`。

```python
from collections import defaultdict


counts = defaultdict(int)
for word in ["red", "blue", "red"]:
    counts[word] += 1
print(dict(counts))

labels = defaultdict(lambda: "Default Value")
labels["a"] = 42
print(labels["a"])
print(labels["missing"])
```

### 分组统计

`Counter` 和 `defaultdict(list)` 经常一起用于统计频次和按条件分组。

```python
from collections import Counter, defaultdict


words = ["ray", "dask", "ray", "celery", "arq", "celery", "ray"]
counts = Counter(words)
print("Most common:", counts.most_common(2))

grouped: defaultdict[str, list[str]] = defaultdict(list)
for tool in words:
    grouped[tool[0]].append(tool)

print("Grouped:", dict(grouped))
```

### 嵌套结构

工厂函数可以递归返回 `defaultdict`，用于构造树形结构。

```python
import json
from collections import defaultdict


def tree():
    return defaultdict(tree)


root = tree()
root["Page"]["Python"]["defaultdict"]["Title"] = "Using defaultdict"
root["Page"]["Python"]["defaultdict"]["Subtitle"] = "Create a tree"
root["Page"]["Java"] = None
print(json.dumps(root, indent=4))
```

## bisect

`bisect` 适合维护已经排序的小到中等规模列表。它的插入成本仍然是 `O(n)`，但查找插入位置是 `O(log n)`，适合读多写少或数据量不大的场景。

```python
import bisect

scores: list[int] = []
for score in [80, 95, 70, 88, 95]:
    bisect.insort(scores, score)

print("Sorted scores:", scores)
print("First >= 88:", bisect.bisect_left(scores, 88))
print("First > 88:", bisect.bisect_right(scores, 88))


def grade_rank(sorted_scores: list[int], target: int) -> int:
    return bisect.bisect_left(sorted_scores, target)


print("Rank index:", grade_rank(scores, 90))
```

## heapq

`heapq` 默认是最小堆，适合实现 Top K、优先级任务调度和流式最值维护。需要最大堆时常见做法是存入负数或反向排序键。

```python
import heapq
from dataclasses import dataclass, field


@dataclass(order=True)
class Task:
    priority: int
    name: str = field(compare=False)


tasks = [Task(3, "send-email"), Task(1, "sync-user"), Task(2, "build-report")]
heapq.heapify(tasks)

while tasks:
    print("Run:", heapq.heappop(tasks))

numbers = [9, 4, 7, 1, 3, 8]
print("Top 3:", heapq.nlargest(3, numbers))
print("Smallest 2:", heapq.nsmallest(2, numbers))
```

## collections.deque

`collections.deque` 是双端队列，头尾插入和弹出都是 `O(1)`。它适合固定长度窗口、生产消费队列和 BFS 队列。

```python
from collections import deque

window: deque[int] = deque(maxlen=3)
for value in [10, 20, 30, 40]:
    window.append(value)
    print("Window:", list(window), "Sum:", sum(window))

queue: deque[str] = deque(["root"])
while queue:
    node = queue.popleft()
    print("Visit:", node)
    if node == "root":
        queue.extend(["left", "right"])
```

## OrderedDict

Python 3.7+ 的普通 `dict` 已经保留插入顺序，但 `OrderedDict` 仍然适合需要移动顺序、实现 LRU 或明确表达“顺序是业务语义”的场景。

```python
from collections import OrderedDict

cache: OrderedDict[str, int] = OrderedDict()
cache["a"] = 1
cache["b"] = 100
cache["c"] = 3

cache.move_to_end("a")
cache["d"] = -1
oldest_key, oldest_value = cache.popitem(last=False)

print("Cache:", cache)
print("Evicted:", oldest_key, oldest_value)
```
