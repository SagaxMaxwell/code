# Lambda 与 map

`lambda` 表达式用于创建匿名函数，适合短小、一次性的表达式。现代代码里，给函数命名和复用时优先使用 `def`；`lambda` 更适合作为 `key`、`map()`、`filter()` 等调用里的内联函数。

## lambda

`lambda` 的主体只能是一个表达式，不能包含多条语句或文档字符串。


```python
def square(x: int) -> int:
    return x * x


def add(x: int = 1, y: int = 2) -> int:
    return x + y


print(square(2))
print(add())
print(add(x=3, y=4))

records = [
    {"name": "Ada", "score": 98},
    {"name": "Grace", "score": 100},
    {"name": "Linus", "score": 95},
]
print(sorted(records, key=lambda item: item["score"], reverse=True))
```

## map

`map()` 把函数惰性地应用到一个或多个可迭代对象上。需要列表时显式调用 `list()`；简单转换通常优先使用列表推导。


```python
def fahrenheit(temp):
    return (9 / 5) * temp + 32


def celsius(temp):
    return (5 / 9) * (temp - 32)


c_temps = [0, 22.5, 40, 100]
f_temps = list(map(fahrenheit, c_temps))
print(f_temps)
print(list(map(celsius, f_temps)))
print([(9 / 5) * temp + 32 for temp in c_temps])
```
