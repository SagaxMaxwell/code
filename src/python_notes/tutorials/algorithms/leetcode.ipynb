### 合并两个有序数组

#### 注意事项

- 空列表
- 数组越界
- 不同数据对象指针

#### 建模

1. 要什么？
   - 输出形式：不返回值，原地修改 `nums1`。
   - 目标标准：`nums1` 的前 `m + n` 个位置包含原 `nums1[:m]` 和 `nums2[:n]` 的全部元素，并且整体非递减。
2. 用什么记？
   - 状态变量：`i` 是 `nums1[:m]` 中下一个待处理元素的位置，`j` 是 `nums2[:n]` 中下一个待处理元素的位置，`result` 是已经合并好的有序前缀。
   - 不变量：`result` 始终保持有序，且正好由已经处理过的元素构成。
3. 怎么变？
   - 转移：比较 `nums1[i]` 和 `nums2[j]`，取较小值追加到 `result`，对应指针右移一位；某一边处理完后，把另一边剩余片段追加到 `result`。
   - 顺序：两个数组都从左到右扫描。
4. 怎么起停？
   - 初始化：`result = []`，`i = j = 0`。
   - 边界 / 终止：`n == 0` 时保留 `nums1[:m]`；`m == 0` 时用 `nums2[:n]` 覆盖 `nums1`。
5. 为什么对？
   - 不变量保持：每次都从两个未处理前缀的头部取较小值，`result` 始终是当前已处理元素里的有序前缀。
   - 覆盖完整：剩余数组本身有序，直接追加仍然有序。

```python
class Solution:
    def merge(self, nums1: list[int], m: int, nums2: list[int], n: int) -> None:
        """
        Do not return anything, modify nums1 in-place instead.
        """
        result = list()
        i, j = 0, 0

        while True:
            if i == m:
                result.extend(nums2[j:n])
                break
            elif j == n:
                result.extend(nums1[i:m])
                break
            elif nums1[i] <= nums2[j]:
                result.append(nums1[i])
                i += 1
            else:
                result.append(nums2[j])
                j += 1

        nums1[:] = result


nums1 = []
m = 0
nums2 = [1, 4]
n = 2

s = Solution()
s.merge(nums1, m, nums2, n)

nums1
```

### 移除元素

#### 注意事项

- 空列表
- 数组越界
- 相邻指针

#### 建模

1. 要什么？
   - 输出形式：返回 `k`，并把不等于 `val` 的元素放到 `nums[:k]`。
   - 目标标准：`nums[:k]` 中没有元素等于 `val`，且 `k` 等于原数组中不等于 `val` 的元素个数。
2. 用什么记？
   - 状态变量：`right` 是当前扫描位置，`left` 是下一个合法元素应该写入的位置，也表示合法前缀长度。
   - 不变量：`nums[:left]` 始终保存已经见过的所有非 `val` 元素。
3. 怎么变？
   - 转移：如果 `nums[right] != val`，执行 `nums[left] = nums[right]`，然后 `left += 1`；否则只移动 `right`。
   - 顺序：`right` 从左到右扫描整个数组。
4. 怎么起停？
   - 初始化：`left = 0`。
   - 边界 / 终止：空数组时循环不执行，返回 `0`。
5. 为什么对？
   - 不变量保持：每个被保留的元素都会进入前缀区，且不会被覆盖回去。
   - 覆盖完整：每个元素只判断一次，所以不会漏保留或误保留。

```python
class Solution:
    def removeElement(self, nums: list[int], val: int) -> int:
        slow = 0

        for fast in range(len(nums)):
            if nums[fast] != val:
                nums[slow], nums[fast] = nums[fast], nums[slow]
                slow += 1

        return slow


nums = [3, 2, 2, 3]
val = 3

s = Solution()
s.removeElement(nums, val)
```

### 删除有序数组中的重复项

#### 注意事项

- 空列表
- 数组越界
- 快慢指针

#### 建模

1. 要什么？
   - 输出形式：返回去重后的长度 `k`，并让 `nums[:k]` 非递减且每个元素只出现一次。
   - 目标标准：`k` 等于原数组中的不同元素个数。
2. 用什么记？
   - 状态变量：`fast` 是当前扫描位置，`slow` 是当前答案前缀最后一个合法元素的位置。
   - 不变量：`nums[:slow + 1]` 始终是已经构造好的合法去重前缀。
3. 怎么变？
   - 转移：如果 `nums[fast] != nums[slow]`，先 `slow += 1`，再把 `nums[fast]` 放到 `nums[slow]`；否则跳过。
   - 顺序：利用数组已经有序的性质，从左到右扫描。
4. 怎么起停？
   - 初始化：非空数组从 `slow = 0` 开始。
   - 边界 / 终止：如果允许空数组，需要先返回 `0`。
5. 为什么对？
   - 不变量保持：有序数组的重复元素一定相邻，只在当前元素不同于答案前缀末尾时加入即可。
   - 覆盖完整：第一次出现的每个不同元素都会被保留，所以结果长度正好是不同元素个数。

```python
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        slow = 0

        for fast in range(len(nums)):
            if nums[fast] != nums[slow]:
                slow += 1
                nums[slow], nums[fast] = nums[fast], nums[slow]

        return slow + 1


nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]

s = Solution()
s.removeDuplicates(nums)
```

### 删除有序数组中的重复项 II

#### 注意事项

- 空列表
- 数组越界
- 快慢指针 + 步长变动
- slow 不是普通指针，而是答案区的长度
- nums[0:slow] 始终是当前已经构造好的合法结果
- 我不是真的删除元素，我只需要把应该保留的元素依次覆盖到数组前面

#### 建模

1. 要什么？
   - 输出形式：返回长度 `k`，并让 `nums[:k]` 非递减且每个数字最多出现两次。
   - 目标标准：`k` 是满足规则后应该保留的元素数量。
2. 用什么记？
   - 状态变量：`fast` 是当前扫描位置，`slow` 是答案前缀长度，也是下一个可写入位置。
   - 不变量：`nums[:slow]` 始终是已经构造好的合法前缀。
3. 怎么变？
   - 转移：如果 `nums[fast] != nums[slow - 2]`，说明当前元素可以加入，写入 `nums[slow]` 后 `slow += 1`；否则跳过。
   - 顺序：`fast` 从左到右扫描有序数组，每个元素判断一次。
4. 怎么起停？
   - 初始化：`n < 3` 时直接返回 `n`；否则 `slow = fast = 2`，默认先保留前两个元素。
   - 边界 / 终止：判断 `nums[slow - 2]` 前必须保证答案区至少有两个元素。
5. 为什么对？
   - 不变量保持：数组有序意味着重复项连续出现，只要当前元素不同于答案前缀倒数第二个元素，就不会产生第三次重复。
   - 覆盖完整：能保留的元素都会被写入前缀，所以答案区长度就是最终保留数量。

```python
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        slow, fast = 2, 2
        n = len(nums)

        if n < 3:
            return n

        while fast < n:
            if nums[fast] != nums[slow - 2]:
                nums[fast], nums[slow] = nums[slow], nums[fast]
                slow += 1
            fast += 1

        return slow


nums = [1, 1, 1, 2, 2, 3]

s = Solution()
s.removeDuplicates(nums)
```

### 多数元素

#### 注意事项

- 哈希表
- `Counter(nums)` 会一次性完成计数

#### 建模

1. 要什么？
   - 输出形式：返回多数元素 `num`。
   - 目标标准：`num` 的出现次数大于 `n // 2`；题目保证多数元素存在，所以只需找出现次数最大的元素。
2. 用什么记？
   - 状态变量：`Counter(nums)` 保存每个元素的真实出现次数，`num` 是当前出现次数最多的元素，`count` 是 `num` 的出现次数。
   - 不变量：`count` 始终是当前已比较元素中的最大频次。
3. 怎么变？
   - 转移：遍历每个 `(元素, 次数)`，如果当前次数 `v >= count`，更新 `num = k`，`count = v`。
   - 顺序：先统计整个数组，再遍历计数字典。
4. 怎么起停？
   - 初始化：`num, count = 0, 0`。
   - 边界 / 终止：LeetCode 约束下数组非空且多数元素存在。
5. 为什么对？
   - 不变量保持：`Counter` 保存了每个元素的真实次数，扫描时始终保留当前最大频次对应的元素。
   - 覆盖完整：多数元素的次数一定最大，所以最终得到的就是多数元素。

```python
from collections import Counter


class Solution:
    def majorityElement(self, nums: list[int]) -> int:
        num, count = 0, 0

        for k, v in Counter(nums).items():
            if v >= count:
                num, count = k, v

        return num


nums = [3, 2, 3]

s = Solution()
s.majorityElement(nums)
```

### 多数元素

#### 注意事项

- 哈希表
- 一边扫描一边更新当前最高频元素

#### 建模

1. 要什么？
   - 输出形式：返回多数元素 `num`。
   - 目标标准：`num` 的出现次数大于 `n // 2`；题目保证多数元素存在，所以维护当前最高频元素即可。
2. 用什么记？
   - 状态变量：`counts` 保存已扫描元素的出现次数，`num` 是当前扫描前缀中出现次数最多的元素，`count` 是 `num` 的出现次数。
   - 不变量：`counts` 始终是当前前缀的精确计数，`num/count` 始终是当前前缀的最高频结果。
3. 怎么变？
   - 转移：`counts[v] += 1`；如果 `counts[v] >= count`，更新 `num = v`，`count = counts[v]`。
   - 顺序：从左到右扫描 `nums`。
4. 怎么起停？
   - 初始化：`counts` 初始化为空计数字典，`num, count = 0, 0`。
   - 边界 / 终止：LeetCode 约束下数组非空且多数元素存在。
5. 为什么对？
   - 不变量保持：扫描任意前缀后，`counts` 都是精确计数，`num/count` 都是当前最高频元素。
   - 覆盖完整：完整扫描后，最高频元素就是题目保证存在的多数元素。

```python
from collections import defaultdict


class Solution:
    def majorityElement(self, nums: list[int]) -> int:
        counts = defaultdict(lambda: 0)
        num, count = 0, 0

        for v in nums:
            counts[v] += 1
            if counts[v] >= count:
                num, count = v, counts[v]

        return num


nums = [3, 2, 3]

s = Solution()
s.majorityElement(nums)
```

### 轮转数组

#### 注意事项

- 数组越界
- 空列表
- 暴力模拟：每次右移一位，重复 `k` 次

#### 建模

1. 要什么？
   - 输出形式：不返回值，原地修改 `nums`。
   - 目标标准：数组元素集合不变，顺序变为整体向右轮转 `k` 位后的结果。
2. 用什么记？
   - 状态变量：`_` 是已经完成的右移轮数，`idx` 是当前正在交换的位置，`nums` 是当前轮转到中间状态的数组。
   - 不变量：每轮结束后，数组都比上一轮多右移了一位。
3. 怎么变？
   - 转移：每次执行 `nums[idx], nums[idx - 1] = nums[idx - 1], nums[idx]`，让当前位置元素向右移动一格；内层循环结束后完成一次右移。
   - 顺序：外层重复 `k` 轮；每轮内层从 `len(nums) - 1` 到 `1` 逆序处理。
4. 怎么起停？
   - 初始化：初始数组就是第 `0` 次轮转结果。
   - 边界 / 终止：`k == 0` 时外层循环不执行；空数组需要单独处理，否则 `k %= len(nums)` 会报错。
5. 为什么对？
   - 不变量保持：一轮从右到左相邻交换，会把最后一个元素逐步换到最前面。
   - 覆盖完整：重复 `k` 次后，数组就恰好轮转 `k` 位。

```python
class Solution:
    def rotate(self, nums: list[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        k %= len(nums)

        for _ in range(k):
            for idx in range(len(nums) - 1, 0, -1):
                nums[idx], nums[idx - 1] = nums[idx - 1], nums[idx]


nums = [1, 2]
k = 3


s = Solution()
s.rotate(nums, k)
nums
```

### 轮转数组

#### 注意事项

- 数组越界
- 空列表
- `k` 需要先对数组长度取模
- 切片赋值要写成 `nums[:] = ...`，否则只是重新绑定局部变量

#### 建模

1. 要什么？
   - 输出形式：不返回值，原地修改 `nums`。
   - 目标标准：数组元素不变，顺序变成向右轮转 `k` 位。
2. 用什么记？
   - 状态变量：`k` 是去掉完整轮转后的有效位移，`nums[-k:]` 是轮转后应该放到前面的尾段，`nums[:-k]` 是轮转后应该放到后面的前段。
   - 不变量：轮转点一旦确定，前后两段的相对顺序都不变。
3. 怎么变？
   - 转移：执行 `nums[:] = nums[-k:] + nums[:-k]`，用目标顺序覆盖原列表内容。
   - 顺序：先规范化 `k`，再一次性拼接切片。
4. 怎么起停？
   - 初始化：先做 `k %= len(nums)`。
   - 边界 / 终止：空数组时 `len(nums) == 0`，需要单独处理，否则取模会报错。
5. 为什么对？
   - 不变量保持：原数组最后 `k` 个元素正好成为新数组前缀，前 `n - k` 个元素成为后缀。
   - 覆盖完整：切片拼接直接构造出目标顺序。

```python
class Solution:
    def rotate(self, nums: list[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        k %= len(nums)
        nums[:] = nums[-k:] + nums[:-k]


nums = [1, 2]
k = 3

s = Solution()
s.rotate(nums, k)
```

### 买卖股票的最佳时机

#### 注意事项

- 数组越界
- 空列表
- 只能买卖一次，必须先买后卖

#### 建模

1. 要什么？
   - 输出形式：返回最大利润 `maxprofit`。
   - 目标标准：合法交易是先在某天买入，再在之后某天卖出；如果没有正利润就返回 `0`，最优标准是利润最大。
2. 用什么记？
   - 状态变量：`minprice` 是到当前天之前或当前天为止见过的最低价格，`maxprofit` 是目前能得到的最大利润。
   - 不变量：`minprice` 始终是扫描前缀里的最小值，`maxprofit` 始终是当前前缀的最优结果。
3. 怎么变？
   - 转移：先用 `price - minprice` 尝试更新 `maxprofit`，再用当前 `price` 尝试更新 `minprice`。
   - 顺序：按时间从左到右扫描价格数组。
4. 怎么起停？
   - 初始化：`maxprofit = 0`，`minprice` 初始化为足够大的数。
   - 边界 / 终止：空数组时循环不执行，返回 `0`。
5. 为什么对？
   - 不变量保持：卖出日固定为当前天时，最优买入日一定是它之前的最低价。
   - 覆盖完整：扫描过程中枚举了每个可能卖出日，所以能保留全局最大利润。

```python
class Solution:
    def maxProfit(self, prices: list[int]) -> int:
        maxprofit = 0
        minprice = 1e9

        for price in prices:
            maxprofit = max(price - minprice, maxprofit)
            minprice = min(price, minprice)

        return maxprofit


prices = [7, 1, 4, 2]

s = Solution()
s.maxProfit(prices)
```

### 买卖股票的最佳时机 II

#### 注意事项

- 空列表
- 数组越界
- 可以多次交易，但同一时间最多持有一股

#### 建模

1. 要什么？
   - 输出形式：返回最大总利润 `maxprofit`。
   - 目标标准：可以完成多笔先买后卖的交易，但同一时间最多持有一股，最优标准是所有合法交易的总利润最大。
2. 用什么记？
   - 状态变量：`minprice` 当前表示上一天价格，`maxprofit` 是累计获得的正收益。
   - 不变量：`maxprofit` 始终保存当前扫描前缀的最优累计收益，`minprice` 始终是下一段比较基准。
3. 怎么变？
   - 转移：`maxprofit += max(0, price - minprice)`，然后 `minprice = price`。
   - 顺序：按时间从左到右扫描。
4. 怎么起停？
   - 初始化：`maxprofit = 0`，`minprice` 初始化为足够大的数。
   - 边界 / 终止：第一天不会产生正收益；空数组时循环不执行，返回 `0`。
5. 为什么对？
   - 不变量保持：任意连续上涨区间的整体利润等于每天正差值之和。
   - 覆盖完整：累加所有相邻正差值，相当于把所有可获得的上涨收益都拿到。

```python
class Solution:
    def maxProfit(self, prices: list[int]) -> int:
        minprice = 1e9
        maxprofit = 0

        for price in prices:
            maxprofit += max(0, price - minprice)
            minprice = price

        return maxprofit


prices = [7, 1, 5, 3, 6, 4]

s = Solution()
s.maxProfit(prices)
```

### 跳跃游戏

#### 注意事项

- 贪心算法
- 空列表
- 数组越界

#### 建模

1. 要什么？
   - 输出形式：返回布尔值，表示能不能到达最后一个下标。
   - 目标标准：如果存在一条路径到达或越过最后一个下标，返回 `True`；否则返回 `False`。
2. 用什么记？
   - 状态变量：`farthest` 是目前扫描过的可达位置能够继续扩展出的最远下标，`last` 是最后一个下标。
   - 不变量：`farthest` 始终代表当前已知可达区间的最右边界。
3. 怎么变？
   - 转移：如果 `i <= farthest`，更新 `farthest = max(farthest, i + nums[i])`；如果 `i > farthest`，直接返回 `False`。
   - 顺序：从左到右扫描数组。
4. 怎么起停？
   - 初始化：`farthest = 0`，`last = len(nums) - 1`。
   - 边界 / 终止：长度为 `1` 时，起点就是终点，应返回 `True`；当前代码在循环内提前判断 `farthest >= last`。
5. 为什么对？
   - 不变量保持：只要当前位置不超过最远可达边界，它就可达并能继续扩展。
   - 覆盖完整：遇到 `i > farthest` 时，中间出现无法跨越的断点，后面都不可达。

```python
class Solution:
    def canJump(self, nums: list[int]) -> bool:
        farthest = 0
        last = len(nums) - 1

        for i, v in enumerate(nums):
            if farthest >= last:
                return True
            elif i > farthest:
                return False
            else:
                farthest = max(farthest, i + v)


nums = [5, 9, 3, 2, 1, 0, 2, 3, 2, 1, 0, 0]

s = Solution()
s.canJump(nums)
```

### 跳跃游戏 II

#### 注意事项

- 贪心算法
- 最少跳数
- 按当前跳数能覆盖的区间分层扫描

#### 建模

1. 要什么？
   - 输出形式：返回到达最后一个下标所需的最少跳数 `jumps`。
   - 目标标准：每次最多向右跳 `nums[i]` 步，在能到达最后一个下标的前提下，让跳跃次数最少。
2. 用什么记？
   - 状态变量：`jumps` 是已经完成的跳跃次数，`end` 是当前这一次跳跃能覆盖到的最右边界，`farthest` 是扫描当前覆盖范围时下一跳能到达的最远位置。
   - 不变量：扫描到当前层内任意位置时，`farthest` 始终表示从当前层所有已扫描位置出发，下一跳能到达的最远边界。
3. 怎么变？
   - 转移：每扫描一个位置 `i`，更新 `farthest = max(farthest, i + nums[i])`；当 `i == end` 时，说明当前这一跳覆盖的区间已经扫完，执行 `jumps += 1`，并令 `end = farthest`。
   - 顺序：从左到右扫描到倒数第二个位置，先扩展下一层最远边界，再在当前层结束时增加跳数。
4. 怎么起停？
   - 初始化：`jumps = 0`，`end = 0`，`farthest = 0`。
   - 边界 / 终止：长度小于 `2` 时不需要跳，返回 `0`；循环只需要处理到 `len(nums) - 2`，因为到达最后一个下标后不需要再跳。
5. 为什么对？
   - 不变量保持：在当前层内扫描所有可达位置时，`farthest` 保留了下一跳的最远可达边界，因此每次跨层都选择了当前跳数下能覆盖的最大范围。
   - 覆盖完整：每次 `i == end` 才增加跳数，等价于 BFS 扫完一层后进入下一层；第一次覆盖到最后位置时，对应的层数就是最少跳数。

```python
class Solution:
    def jump(self, nums: list[int]) -> int:
        end = 0
        farthest = 0
        jumps = 0

        for i, v in enumerate(nums[:-1]):
            farthest = max(farthest, i + v)
            if i == end:
                jumps += 1
                end = farthest

        return jumps


nums = [1]

s = Solution()
s.jump(nums)
```

### H 指数

#### 注意事项

- 空列表
- 计数排序
- 当前写法是枚举每个可能的 `h`，再统计引用数至少为 `h` 的论文数量

#### 建模

1. 要什么？
   - 输出形式：返回 H 指数 `h`。
   - 目标标准：至少有 `h` 篇论文的引用次数大于等于 `h`，并且 `h` 尽量大。
2. 用什么记？
   - 状态变量：`counts[num]` 是引用次数至少为 `num` 的论文数量，`num` 是当前找到的最好 H 指数候选，`ref` 是当前候选对应的论文数量。
   - 不变量：每个候选 `num` 的可行性都由真实统计结果决定，当前答案始终是已检查候选中的最大可行值。
3. 怎么变？
   - 转移：枚举候选值 `num`，统计引用次数 `ref >= num` 的论文数量；在所有满足 `num <= count` 的候选值中选择最大的 `num`。
   - 顺序：先枚举候选 H 指数 `1..n` 并统计，再扫描统计结果选择答案。
4. 怎么起停？
   - 初始化：`counts` 初始化为空计数字典，`num = 0` 表示默认 H 指数为 0。
   - 边界 / 终止：空数组时没有候选值，返回 `0`。
5. 为什么对？
   - 不变量保持：每个候选值是否满足定义都被真实统计出来，答案只在候选可行时更新。
   - 覆盖完整：H 指数只可能在 `0..n` 之间，枚举这些候选并取最大值不会漏掉正确答案。

```python
class Solution:
    def hIndex(self, citations: list[int]) -> int:
        counts = defaultdict(lambda: 0)

        for num in range(1, len(citations) + 1):
            for ref in citations:
                if ref >= num:
                    counts[num] += 1

        num, ref = 0, 1e9

        for k, v in counts.items():
            if k <= v and k - v > num - ref:
                num, ref = k, v

        return num


citations = [1, 1, 3]

s = Solution()
s.hIndex(citations)
```

### H 指数

#### 注意事项

- 空列表
- 排序
- 这个写法先把引用次数降序排好，再顺着看第几篇论文还能支撑第几个 H 指数

#### 建模

1. 要什么？
   - 输出形式：返回 H 指数 `h`。
   - 目标标准：至少有 `h` 篇论文的引用次数大于等于 `h`，并且 `h` 尽量大。
2. 用什么记？
   - 状态变量：`refs` 是按引用次数从大到小排序后的数组，`i` 是当前遍历到的论文编号（从 1 开始），`v` 是当前论文的引用次数，`h` 是当前已经确认的最大可行值。
   - 不变量：在处理到第 `i` 篇时，只要前 `i` 篇都满足 `refs[i - 1] >= i`，当前答案就至少是 `i`；一旦出现 `refs[i - 1] < i`，说明 `i` 已经不可行，答案只能停在 `i - 1`。
3. 怎么变？
   - 转移：按 `i = 1..n` 依次检查 `refs[i - 1]` 和 `i` 的关系。
   - 如果 `v == i`，说明刚好有 `i` 篇论文都至少被引用 `i` 次，`i` 已经是可行答案，可以直接返回。
   - 如果 `v < i`，说明连第 `i` 篇都达不到 `i` 次引用，`i` 及之后都不可能更大，直接返回 `i - 1`。
   - 如果 `v > i`，说明当前仍然可行，把 `h` 更新为 `i`，继续向后看。
4. 怎么起停？
   - 初始化：`refs = sorted(citations, reverse=True)`，`h = 0`。
   - 边界 / 终止：空列表时循环不执行，返回 `0`；遍历结束仍未提前返回时，说明所有论文都满足 `refs[i - 1] >= i`，答案就是最后一次更新的 `h`。
5. 为什么对？
   - 排序后，越往后引用次数越少，所以可行性是单调下降的。
   - 一旦出现 `refs[i - 1] < i`，后面的论文只会更差，答案不可能再变大。

```python
class Solution:
    def hIndex(self, citations: list[int]) -> int:
        refs = sorted(citations, reverse=True)
        h = 0

        for i, v in enumerate(refs, start=1):
            if i == v:
                return i
            elif i > v:
                return max(i - 1, 0)
            else:
                h = i

        return h


citations = [100]

s = Solution()
s.hIndex(citations)
```

### O(1) 时间插入、删除和获取随机元素

#### 注意事项

- 数组
- 索引
- 字典

```python
from secrets import choice


class RandomizedSet:
    def __init__(self):
        self._idxs: dict[int, int] = dict()
        self._vals: list[int] = list()

    def insert(self, val: int) -> bool:
        if val in self._idxs:
            return False

        self._idxs[val] = len(self._vals)
        self._vals.append(val)
        return True

    def remove(self, val: int) -> bool:
        if val in self._idxs:
            idx = self._idxs[val]
            self._vals[-1], self._vals[idx] = self._vals[idx], self._vals[-1] 
            self._idxs[self._vals[idx]] = idx
            self._idxs.pop(val) 
            self._vals.pop()
            return True

        return False

    def getRandom(self) -> int:
        return choice(self._vals)


obj = RandomizedSet()
obj.remove(0)
obj.remove(0)
obj.insert(0)
obj.getRandom()
obj.remove(0)
obj.insert(0)
```

### 除了自身以外数组的乘积

#### 注意事项

- 元素为 0
- 前后各自迭代，计算前缀/后缀乘积

```python
class Solution:
    def productExceptSelf(self, nums: list[int]) -> list[int]:
        n = len(nums)
        pre_production = [1] * n
        suf_production = [1] * n
        answer = [1] * n

        for i in range(1, n):
            pre_production[i] = pre_production[i - 1] * nums[i - 1]

        for i in range(n - 2, -1, -1):
            suf_production[i] = suf_production[i + 1] * nums[i + 1]

        for i in range(n):
            answer[i] = pre_production[i] * suf_production[i]

        return answer


nums = [-1, 1, 0, -3, 3]

s = Solution()
s.productExceptSelf(nums)
```

### 除了自身以外数组的乘积

#### 注意事项

- 元素为 0
- 前后各自迭代，计算前缀/后缀乘积

```python
class Solution:
    def productExceptSelf(self, nums: list[int]) -> list[int]:
        n = len(nums)
        answer = [1] * n

        for i in range(1, n):
            answer[i] = answer[i - 1] * nums[i - 1]
        
        r_prod = 1

        for i in reversed(range(n)):
            answer[i] *= r_prod 
            r_prod *= nums[i]
            
        return answer


[1, 2, 3, 4]
[1, 1, 2, 6]

nums = [-1, 1, 0, -3, 3]

s = Solution()
s.productExceptSelf(nums)
```

### 加油站

```python
class Solution:
    def canCompleteCircuit(self, gas: list[int], cost: list[int]) -> int:
        total = 0
        start = 0
        tank = 0

        for i in range(len(gas)):
            diff = gas[i] - cost[i]
            tank += diff
            total += diff

            if tank < 0:
                start = i + 1
                tank = 0

        return start if total > -1 else -1


gas = [1, 2, 3, 4]
cost = [2, 3, 4, 5]

s = Solution()
s.canCompleteCircuit(gas, cost)
```

### 分发糖果

#### 建模

1. 要什么？
   - 输出形式：返回满足规则时需要发出的最少糖果总数 `ret`。
   - 目标标准：每个孩子至少拿到 `1` 个糖果，且评分更高的孩子必须比相邻评分更低的孩子拿到更多糖果，总数最小。
2. 用什么记？
   - 状态变量：`left[i]` 是只看左边关系时第 `i` 个孩子需要的最少糖果数，`right[i]` 是只看右边关系时第 `i` 个孩子需要的最少糖果数，`ret` 是合并后的总数。
   - 不变量：`left` 和 `right` 分别保存单侧约束下的最小可行值。
3. 怎么变？
   - 转移：从左到右时，若 `ratings[i] > ratings[i - 1]`，则 `left[i] = left[i - 1] + 1`；从右到左时，若 `ratings[i] > ratings[i + 1]`，则 `right[i] = right[i + 1] + 1`；最后取 `max(left[i], right[i])`。
   - 顺序：第一遍从左到右，第二遍从右到左，第三遍从左到右合并左右两边的最低要求。
4. 怎么起停？
   - 初始化：`n = len(ratings)`，`left = [1] * n`，`right = [1] * n`，第三遍初始化 `ret = 0`。
   - 边界 / 终止：空数组时三个循环都不会执行，返回 `0`；题目通常保证 `ratings` 非空。
5. 为什么对？
   - 不变量保持：相邻约束可以拆成左约束和右约束，两次扫描分别得到满足单侧约束的最小值。
   - 覆盖完整：每个孩子必须同时满足两侧约束，所以取二者最大值就是该位置的最小合法糖果数，总和最小。

```python
class Solution:
    def candy(self, ratings: list[int]) -> int:
        n = len(ratings)

        left = [1] * n
        right = [1] * n

        for i in range(1, n):
            if ratings[i] > ratings[i - 1]:
                left[i] = left[i - 1] + 1

        for i in reversed(range(n-1)):
            if ratings[i] > ratings[i + 1]:
                right[i] = right[i + 1] + 1

        ret = 0
        for i in range(n):
            ret += max(left[i], right[i])

        return ret
```

### 罗马数字转整数

```python
class Solution:
    def __init__(self):
        self._codes = {
            "I": 1,
            "V": 5,
            "X": 10,
            "L": 50,
            "C": 100,
            "D": 500,
            "M": 1000,
        }

    def romanToInt(self, s: str) -> int:
        result = 0
        latest = 0

        for c in s[::-1]:
            val = self._codes[c]
            if val < latest:
                result -= val
            else:
                result += val
            latest = val

        return result


s = "MCMXCIV"

obj = Solution()
obj.romanToInt(s)
```

### 整数转罗马数字

```python
class Solution:
    def __init__(self):
        self._codes = (
            (1000, "M"),
            (900, "CM"),
            (500, "D"),
            (400, "CD"),
            (100, "C"),
            (90, "XC"),
            (50, "L"),
            (40, "XL"),
            (10, "X"),
            (9, "IX"),
            (5, "V"),
            (4, "IV"),
            (1, "I"),
        )

    def intToRoman(self, num: int) -> str:
        romans = list()

        for val, symbol in self._codes:
            while num >= val:
                num -= val
                romans.append(symbol)
            if num <= 0:
                break

        return "".join(romans)


s = Solution()
s.intToRoman(175)
```

### Z 字型变换

```python
def convert(s: str, numRows: int) -> str:
    if numRows == 1 or numRows >= len(s):
        return s

    cycle = 2 * numRows - 2
    res = []

    for r in range(numRows):
        for i in range(0, len(s), cycle):
            if i + r < len(s):
                res.append(s[i + r])

            if 0 < r < numRows - 1 and i + cycle - r < len(s):
                res.append(s[i + cycle - r])

    return ''.join(res)
```

### Z 字型变换

```python
class Solution:
    def convert(self, s: str, numRows: int) -> str:
        if numRows < 2:
            return s

        result: list[tuple[int, int, str]] = list()
        row = -1
        col = 0

        for i, c in enumerate(s):
            x, y = divmod(i, numRows - 1)

            if (x > 0) and (x % 2 == 1) and (y > 0):
                col += 1
                row -= 1
            elif (x > 0) and (x % 2 == 0) and (y == 0):
                col += 1
                row -= 1
            else:
                row += 1

            result.append((row, col, c))

        return "".join(tup[-1] for tup in sorted(result))


s = "PAYPALISHIRING"
numRows = 4

obj = Solution()
obj.convert(s, numRows)
```

### 文本左右对齐

```python
from dataclasses import dataclass, field


@dataclass()
class Row:
    count: int = 0
    length: int = 0
    words: list[str] = field(default_factory=lambda: list())

    @property
    def adjust_width(self) -> int:
        return max(self.count - 1, 0) + self.length

    def add(self, w: str) -> None:
        self.count += 1
        self.length += len(w)
        self.words.append(w)

    def adjust(self, maxWidth: int) -> str:
        result = ""

        if self.count < 2:
            result = self.words[0].ljust(maxWidth)
        else:
            a, b = divmod(maxWidth - self.length, self.count - 1)
            for w in self.words[:-1]:
                result += w + " " * a
                if b > 0:
                    result += " "
                    b -= 1
            result += self.words[-1]

        return result


class Solution:
    def fullJustify(self, words: list[str], maxWidth: int) -> list[str]:
        row = Row()
        result: list[str] = list()

        for w in words:
            if sum((row.count, row.length, len(w))) > maxWidth:
                result.append(row.adjust(maxWidth))
                row = Row()
            row.add(w)

        result.append(" ".join(row.words).ljust(maxWidth))

        return result


words = ["What", "must", "be", "acknowledgment", "shall", "be"]
maxWidth = 16

s = Solution()
s.fullJustify(words, maxWidth)
```

### 同构字符串

```python
class Solution:
    def isIsomorphic(self, s: str, t: str) -> bool:
        ab = dict()
        ba = dict()

        for a, b in zip(s, t):
            if a not in ab:
                if b not in ba:
                    ab[a] = b
                    ba[b] = a
                else:
                    return False
            else:
                if b not in ba:
                    return False
                elif ab[a] != b:
                    return False

        return True


s = "e1e"
t = "add"

obj = Solution()
obj.isIsomorphic(s, t)
```

### 两数之和

#### 建模

1. 要什么？
   - 输出形式：返回两个数之和等于 `target` 的原数组下标。
   - 目标标准：两个下标不同，且对应元素之和等于 `target`。
2. 用什么记？
   - 状态变量：`arr` 保存 `(原下标, 数值)` 的排序结果，`a` 和 `b` 是排序数组上的左右指针。
   - 不变量：`arr` 按数值有序，左右指针之间保留当前还可能组成答案的候选区间。
3. 怎么变？
   - 转移：如果 `arr[a][1] + arr[b][1] == target`，返回两个原下标；如果和太大，`b -= 1`；如果和太小，`a += 1`。
   - 顺序：先排序，再用双指针从两端向中间扫描。
4. 怎么起停？
   - 初始化：`a = 0`，`b = len(nums) - 1`。
   - 边界 / 终止：指针不能交叉；当前题通常保证存在答案。
5. 为什么对？
   - 不变量保持：排序后，如果当前和太大，只能左移右指针来减小和；如果当前和太小，只能右移左指针来增大和。
   - 覆盖完整：双指针按单调方向排除不可能的组合，`arr` 保留了原下标，所以能返回题目要求的结果。

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        arr = sorted(enumerate(nums), key=lambda x: x[-1])
        a, b = 0, len(nums) - 1

        while (a < len(nums) - 1) and (b > 0):
            if (twosum := arr[a][-1] + arr[b][-1]) == target:
                return [arr[a][0], arr[b][0]]
            elif twosum > target:
                b -= 1
            else:
                a += 1


nums = [2, 7, 11, 15]
target = 9

s = Solution()
s.twoSum(nums, target)
```

### 两数之和

#### 建模

1. 要什么？
   - 输出形式：返回两个数之和等于 `target` 的下标。
   - 目标标准：两个下标不同，且对应元素之和等于 `target`。
2. 用什么记？
   - 状态变量：`hashtable` 保存已经扫描过的数值到下标的映射。
   - 不变量：扫描到位置 `i` 时，哈希表只包含 `i` 左边的元素。
3. 怎么变？
   - 转移：扫描当前 `num` 时，先查 `target - num` 是否已经出现；如果出现，直接返回两个下标；否则把当前数写入哈希表。
   - 顺序：从左到右扫描，并在迭代过程中动态构建哈希表。
4. 怎么起停？
   - 初始化：`hashtable = {}`。
   - 边界 / 终止：当前元素不能和自己配对，所以要先查补数，再写入当前元素。
5. 为什么对？
   - 不变量保持：每次查询时，哈希表只包含之前出现过的元素，因此不会把当前元素和自己配对。
   - 覆盖完整：每个元素都会作为第二个数被检查一次，若补数存在就能立刻找到答案。

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        hashtable = dict()
        for i, num in enumerate(nums):
            if target - num in hashtable:
                return [hashtable[target - num], i]
            hashtable[nums[i]] = i
        return []
```

### 快乐数

#### 建模

1. 要什么？
   - 输出形式：返回布尔值，判断一个数是不是快乐数。
   - 目标标准：不断替换成各位数字平方和后，最终到达 `1` 返回 `True`；进入循环返回 `False`。
2. 用什么记？
   - 状态变量：`get_total(n)` 计算当前数字的各位平方和，`total` 是下一轮要判断的数字。
   - 不变量：每一轮的 `total` 都是上一个数字唯一确定的后继状态。
3. 怎么变？
   - 转移：把 `n` 替换成 `get_total(n)`；如果结果是 `1`，返回 `True`；如果结果是 `4`，返回 `False`；否则递归继续。
   - 顺序：每轮只依赖上一轮得到的数字。
4. 怎么起停？
   - 初始化：先计算 `total = get_total(n)`。
   - 边界 / 终止：`1` 是快乐数终点；`4` 是非快乐数循环中的代表点。
5. 为什么对？
   - 不变量保持：各位平方和转换是确定的，所以同一个数字之后的路径完全固定。
   - 覆盖完整：快乐数路径最终到达 `1`，非快乐数路径会进入包含 `4` 的固定循环。

```python
class Solution:
    def get_total(self, n: int) -> list[int]:
        result = 0

        while n > 0:
            n, m = divmod(n, 10)
            result += m**2

        return result

    def isHappy(self, n: int) -> bool:
        total = self.get_total(n)

        match total:
            case 1:
                return True
            case 4:
                return False
            case _:
                return self.isHappy(total)


s = Solution()
s.isHappy(1921)
```

### 快乐数

```python
class Solution:
    def get_total(self, n: int) -> list[int]:
        result = 0

        while n > 0:
            n, m = divmod(n, 10)
            result += m**2

        return result

    def isHappy(self, n: int) -> bool:
        sad = {4, 16, 37, 58, 89, 145, 42, 20}
        total = self.get_total(n)

        if total in sad:
            return False
        elif total == 1:
            return True
        else:
            return self.isHappy(total)


s = Solution()
s.isHappy(19)
```

### 快乐数

```python
class Solution:
    def get_total(self, n: int) -> list[int]:
        result = 0

        while n > 0:
            n, m = divmod(n, 10)
            result += m**2

        return result

    def isHappy(self, n: int) -> bool:
        sad = set()

        while n != 1 and n not in sad:
            sad.add(n)
            n = self.get_total(n)

        return n == 1


s = Solution()
s.isHappy(19)
```

### 存在重复元素 II

```python
class Solution:
    def containsNearbyDuplicate(self, nums: list[int], k: int) -> bool:
        counter = dict()

        for i, v in enumerate(nums):
            if (v in counter) and (abs(counter[v] - i) <= k):
                return True
            else:
                counter[v] = i

        return False


nums = [1, 0, 0, 1]
k = 0

s = Solution()
s.containsNearbyDuplicate(nums, k)
```
